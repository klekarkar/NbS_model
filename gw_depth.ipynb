#### Groundwater flow Equations 

Darcy's velocity

$v_{Rlocal}(t) = \frac{K_{hGWlocal}}{\phi} i_{local} (t)$

Kh= characteristic hydraulic conductivity to calculate subsurface local groundwater flow  
ilocal: time dependent hydraulic gradient

$i_{local}(t) = \Delta h (t)/{L_{GWlocal}}$

$L_{GWlocal}$ is distance between upstream and downstream boundaries  
$\Delta h $ time varying head difference between up and downstream


the response time is given by:  

$t_{Rlocal}(t) = L_{GWlocal}/v_{Rlocal}(t) $  

The local subsurface flow component that feeds the regional groundwater

$q_{local}(t) = \Delta h (t) /t_{Rlocal}(t) $


Regional hydraulic conductivity:  

$Q_{reg}(t) = \frac {K_{hGWreg}}{\phi} i_{reg} (t) $


$i_{reg} (t) =  \Delta h (t) /L_{GWReg}$

### Soil Water Balance Equations

Potential Evaporation

$\lambda E_{p} = \alpha * \frac{\Delta}{\Delta + \gamma} * (R_{n} - G) $

a constant value of $\alpha$ = 0.8 is used to parameterise the tall
canopy fraction, while a value of $\alpha$ =1.26 is applied in both
the short vegetation and bare soil fractions.

----

Actual Evapotranspiration

$E = S*E_{p} + I - \beta*I$ 

 I = interception loss
 S = Evaporation stress factor (1= No stress, 0 = Maximum stress)

$\beta$ =0.07 – Gash and Stewart,1977)

----

Evaporative stress  
Tall canopy:  

$S = 1 -(\frac{\omega_{c} - \omega_{w}}{\omega_{c} - \omega_{wp}}) ^ 2$

Short vegetation:  

$S = 0.5 * (1 - \sqrt(\frac{\omega_{c} - \omega_{w}}{\omega_{c} - \omega_{wp}}) + \frac{\tau}{0.8})$

Bare Soil:  

$S = 1 - \sqrt(\frac{\omega_{c} - \omega_{w}}{\omega_{c} - \omega_{wp}})$

$\tau$ is used in short-vegetated covers to account for the effect of seasonal or occasional changes in
biomass content on evaporation

$\tau$ is given at two vegetation densities and is equal to 0.2 or 0.8

-----

Soil water balance:  

$\theta_{time} = \theta_{time - 1} + \frac{F^{l -1} - E^{(l)}_{time -1} - F^{(l)}_{time}}{\Delta z ^{(l)}}$

Percolation:  

$F^{l}_{i} = (\theta^{(l)}_{time} - \theta _{(fc)}) \Delta Z$
